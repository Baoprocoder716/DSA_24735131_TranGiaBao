```python
#In ra danh sách sản phẩm:
def xuatDanhSach():
    print("Danh sach san pham cua toi la:")
    print("\n----\n")
    for p in products:
        print(f"Ma san pham: {p['id']}")
        print(f"Ten san pham: {p['name']}")
        print(f"Gia san pham: {p['price']}")
        print(f"Mo ta san pham: {p['description']}")
        print("\n----\n")
xuatDanhSach()
```

```python
#Thêm 1 sản phẩm
def themSP():
    id = products[-1]["id"] + 1 if products else 1
    ten = str(input("Nhap ten san pham: "))
    gia = int(input("Nhap gia san pham: "))
    mota = input("Nhap mo ta san pham: ")
    sp = {
            "id": id,
            "name": ten,
            "price": gia,
            "description": mota
        }
    products.append(sp)
themSP()
```

```python
#Thêm nhiều sản phẩm
def themNhieuSanPham():
    n = int(input("Hay nhap so luong ban muon nhap."))
    for x in (0,n):
        themSP()
themNhieuSanPham()
```

```python
#Cập nhật sản phẩm
def capNhat(id):
    found = False
    for p in products:
        if p['id'] == id:
            name = input("Nhap ten moi.")
            price = input("Nhap gia moi.")
            description = input("Nhap mo ta moi.")
            
            if name:
                p['name'] = name
            if price:
                p['price'] = price
            if description:
                p['description'] = description
            
            print(f"Cap nhat thanh cong {id}.")

if not found:
    print(f"Khong tim thay {id}.")
```
