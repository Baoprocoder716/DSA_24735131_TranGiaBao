```python
from abc import ABC, abstractmethod
class sinhVien(ABC):
    uni = "Trường Đại học Công nghiệp TP.HCM"
    def __init__(self, mssv, hoten):
        self.mssv = mssv
        self.hoten = hoten
        self.__cccd = '123'

    def get_cccd(self):
        return self.__cccd
    
    def set_cccd(self, cccd):
        self.__cccd = cccd

    def InThongTin(self):
        print(f"{self.mssv} - {self.hoten} - {self.uni}")
    
    @abstractmethod
    def tinhDiemTrungBinh(self):
        pass
```

```python
sv1 = sinhVien('123456', 'Hien')
sv2 = sinhVien('123455', 'Tam')
```

```python
sv1.InThongTin()
sv2.InThongTin()
```

```python
#Overiding
class KSTN(sinhVien):
    def __init__(self, mssv, hoten, diemCTRR, diemTA):
        super().__init__(mssv, hoten)
        self.diemCTRR = diemCTRR
        self.diemTA = diemTA

    def InThongTin(self):
        super().InThongTin()
        print(f"Diem CTRR la: {self.diemCTRR} - Diem TA la: {self.diemTA}")
    
    def tinhDiemTrungBinh(self):
        dtb = (self.diemCTRR + self.diemTA) / 2
        print(f"Diem trung binhh: {dtb}")
```

```python
sv_KSTN1 = KSTN('001', 'Trần Gia Bảo', 8, 7)
sv_KSTN1.InThongTin()
```

```python
class TCTA(sinhVien):
    def __init__(self, mssv, hoten, diemTA3, diemTA4):
        super().__init__(mssv, hoten)
        self.diemTA3 = diemTA3
        self.diemTA4 = diemTA4

    def InThongTin(self):
        super().InThongTin()
        print(f"Diem TA3 la: {self.diemTA3} - Diem TA4 la: {self.diemTA4}")

    def tinhDiemTrungBinh(self):
        dtb1 = (self.diemTA3 + self.diemTA4) / 2
        print(f"Diem trung binh la: {dtb1}")
```

```python
sv_TCTA1 = TCTA('002', 'Trần Gia Bảo', 8.5, 9)
sv_TCTA1.InThongTin()
```
