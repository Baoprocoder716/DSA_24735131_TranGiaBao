```python
from abc import ABC, abstractmethod
class person(ABC):
    def __init__(self, id, name):
        self._id = id
        self._name = name

    def get_id(self):
        return self.id
    def get_name(self):
        return self.name
    
    def set_name(self, name):
        self._name = name
    def set_id(self, id):
        self._id = id
        
    @abstractmethod
    def get_role(self):
        pass
    @abstractmethod
    def to_dict(self):
        pass
```

```python
class student(person):
    def __init__(self, id, name, gpa):
        super().__init__(id, name)
        self.__gpa = gpa #private
    
    def set_gpa(self, gpa_new):
        self.__gpa = gpa_new
    def get_gpa(self):
        return self.__gpa
    
    def get_role(self):
        return "Student"
    def to_dict(self):
        return {
            "id": self.get_id(),
            "name": self.get_name(),
            "gpa": self.get_gpa()
        }
```

```python
class UnderStudent(student):
    def __init__(self, id, name, gpa, year):
        super().__init__(id, name, gpa)
        self._year = year
    
    def set_year(self, year):
        self._year = year
    def get_year(self):
        return self._year
    
    def get_role(self):
        return "Undergraduate students"
    def to_dict(self):
        return {
            "id": self._id,
            "name": self._name,
            "gpa": self.__gpa,
            "year": self._year
        }
```

```python
class GraduateStudent(student):
    def __init__(self, id, name, gpa, thesis):
        super().__init__(id, name, gpa)
        self._thesis = thesis
    
    def set_thesis(self, thesis):
        self._thesis = thesis
    def get_thesis(self):
        return self._thesis
    
    def get_role(self):
        return "Graduates"
    def to_dict(self):
        return {
            "id": self._id,
            "name": self._name,
            "gpa": self.__gpa,
            "thesis": self._thesis
        }
```

```python
class Manager:
    def __init__(self):
        self._lstStud = []

    #Them sinh vien
    def add_student(self, s):
        self._lstStud.append(s)
    
    #Tim sinh vien
    def find_student(self, id):
        for n in self._lstStud:
            if n.get_id() == id:
                return n
        return None
    
    #Xoa sinh vien
    def remove_student(self, id):
        s = self.find_student(id)
        if s:
            del self._lstStud[s]
            return s
        return None
    
    #Cap nhat sinh vien
    def update_student(self, id, name, gpa):
        s = self.find_student(id)
        if s:
            if name:
                s.set_name() = name
            if gpa:
                s.set_gpa() = gpa
                
```
