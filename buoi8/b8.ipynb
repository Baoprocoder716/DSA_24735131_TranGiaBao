```python
class Node:
    def __init__(self, data):
        self.data = data
        self.children = []
```

```python
class Tree:
    def __init__(self):
        self.root = None
    
    def add_node(self, parent, data):
        node = Node(data)
        if parent is None:
            self.root = node
        else:
            self.root.children.append(node)
        return node
    
    def NLR(self, node = None):
        if node is None:
            node = self.root
        if node:
            print(node.data, end=" ")
            for c in node.children:
                self.NLR(c)
    
    def LRN(self, node = None):
        if node is None:
            node = self.root
        if node:
            for c in node.children:
                self.LRN(c)
            print(node.data, end=" ")
    
    def RLN(self, node = None):
        if node is None:
            node = self.root
        if node:
            for c in reversed(node.children):
                self.RLN(c)
            print(node.data, end=" ")
```

```python
class Node:
    def __init__(self, Data):
        self.Data = Data
        self.left = None
        self.right = None

    def add_node(self, value):
        if value < self.Data:
            if self.left is None:
                self.left = Node(value)
            else:
                self.left.add_node(value)
        elif value > self.Data:
            if self.right is None:
                self.right = Node(value)
            else:
                self.right.add_node(value)
    
    def NLR(self):
        print(self.Data, end=" ")
        if self.left:
            self.left.NLR()
        if self.right:
            self.right.NLR()
    
    def LNR(self):
        if self.left:
            self.left.LNR()
        print(self.Data, end=" ")
        if self.right:
            self.right.LNR()
    
    def RLN(self):
        if self.right:
            self.right.RLN()
        if self.left:
            self.left.RLN()
        print(self.Data, end=" ")
    
    def Tim_Kiem(self, value):
        if value == self.Data:
            return True
        if self.left:
            self.left.Tim_Kiem(value)
        if self.right:
            self.right.Tim_Kiem(value)
        return False
    
    def Dem(self):
        count = 1
        if self.left:
            count += self.left.Dem()
        if self.right:
            count += self.right.Dem()
        return count
```

```python
lst = [5,8,1,3,2]
n = Node(lst[0])
for i in lst[1:]:
    n.add_node(i)
```
