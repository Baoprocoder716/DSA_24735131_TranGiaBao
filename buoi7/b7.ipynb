```python
class Book:
    def __init__(self, ma, ten, tac_gia, gia = None, nam_xb = None):
        self.ma = ma
        self.ten = ten
        self.tac_gia = tac_gia
        self.gia = gia
        self.nam_xb = nam_xb
    
    def InThongTin(self):
        return f"Ma sach: {self.ma}, Ten sach: {self.ten}, Tac gia: {self.tac_gia}, Gia sach: {self.gia}, Nam xuat ban: {self.nam_xb}"
    
    def ThoiGianXuatBan(self):
        pass
```

```python
class SachGiaoKhoa(Book):
    def __init__(self, ma, ten, tac_gia, gia=None, nam_xb=None, lop = None, mon = None):
        super().__init__(ma, ten, tac_gia, gia, nam_xb)
        self.lop = lop
        self.mon = mon

    def InThongTin(self):
        return f"super().InThongTin(), Lop: {self.lop}, Mon hoc: {self.mon}"
    
    def ThoiGianXuatBan(self):
        print("Sach giao khoa xuat ban thang 9/26")
```

```python
class SachThamKhao(Book):
    def __init__(self, ma, ten, tac_gia, gia=None, nam_xb=None, chuyen_nganh = None, cap_do = None):
        super().__init__(ma, ten, tac_gia, gia, nam_xb)
        self.chuyen_nganh = chuyen_nganh
        self.cap_do = cap_do

    def InThongTin(self):
        return f"super().InThongTin(), Chuyen nganh: {self.chuyen_nganh}, Cap do: {self.cap_do}"
    
    def ThoiGianXuatBan(self):
        print("Sach tham khao xuat ban thang 3,4,5/2026")
```

```python
from abc import ABC, abstractmethod
class BaseBook(ABC):
    @abstractmethod
    def __init__(self):
        pass

    @abstractmethod
    def push(self, book):
        pass

    @abstractmethod
    def pop(self):
        pass
```

```python
class StackBook(BaseBook):
    def __init__(self):
        self.lstStack = []
    
    def is_empty(self):
        return len(self.lstStack) == 0
    
    def push(self, book):
        self.lstStack.append(book)

    def pop(self):
        return None if self.lstStack.is_empty() else self.lstStack.pop()
```

```python
class QueueBook(BaseBook):
    def __init__(self):
        self.lstQueue = []

    def is_empty(self):
        return len(self.lstQueue) == 0
    
    def push(self, book):
        self.lstQueue.append(book)

    def pop(self):
        return None if self.lstQueue.is_empty() else self.lstQueue.pop(0)
```
