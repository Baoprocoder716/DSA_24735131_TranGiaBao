```python
#Quick sort:
#Lưu ý: mid = pivot(Điểm tựa), left(Những số bé hơn pivot), right(Những số lớn hơn pivot)
```

```python
def quick_sort(arr):
    if len(arr) <= 1:
        return arr
    
    pivot = arr[len(arr) // 2]

    left = [x for x in arr if x < pivot]
    mid = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]

    return quick_sort(left) + mid + quick_sort(right)

arr = [15, 1, 23, 4, 6]
quick_sort(arr)
```

```python
from abc import ABC, abstractmethod
class BaseEm(ABC):
    @abstractmethod
    def is_empty(self):
        pass
    
    @abstractmethod
    def push(self, nv):
        pass

    @abstractmethod
    def pop(self):
        pass
```

```python
class StackEm(BaseEm):
    def __init__(self):
        self.dsStack = []

    def is_empty(self):
        return len(self.dsStack) == 0
    
    def push(self, nv):
        self.dsStack.append(nv)

    def pop(self):
        return None if self.dsStack.is_empty() else self.dsStack.pop()
```

```python
class QueueEm(BaseEm):
    def __init__(self):
        self.dsQueue = []

    def is_empty(self):
        return len(self.dsQueue) == 0
    
    def push(self, nv):
        self.dsQueue.append(nv)

    def pop(self):
        return None if self.dsQueue.is_empty() else self.dsQueue.pop(0)
```
