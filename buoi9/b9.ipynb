```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None
    
    def them_node(self, value):
        if value < self.data:
            if self.left is None:
                self.left = Node(value)
            else:
                self.left.them_node(value)
        elif value > self.data:
            if self.right is None:
                self.right = Node(value)
            else:
                self.right.them_node(value)
    
    def NLR(self):
        print(self.data, end=" ")
        if self.left:
            self.left.NLR()
        if self.right:
            self.right.NLR()

    def RLN(self):
        if self.right:
            self.right.RLN()
        if self.left:
            self.left.RLN()
        print(self.data, end=" ")
    
    def LRN(self):
        if self.left:
            self.left.LRN()
        if self.right:
            self.right.LRN()
        print(self.data, end=" ")
    
    def count_leaf(self):
        if not self.left and not self.right:
            return 1
        count = 0
        if self.left:
            count += self.left.count_leaf()
        if self.right:
            count += self.right.count_leaf()
        return count
    
    def count_one_node(self):
        if (self.left and not self.right) or (not self.left and self.right):
            return 1
        count = 0
        if self.left:
            count += self.left.count_one_node()
        if self.right:
            count += self.right.count_one_node()
        return count
    
    def count_two_node(self):
        if self.left and self.right:
            return 1
        count = 0
        if self.left:
            count += self.left.count_two_node()
        if self.right:
            count += self.right.count_two_node()
        return count
    
    def search(self, value):
        if value == self.data:
            return True
        if self.left:
            self.left.search(value)
        if self.right:
            self.right.search(value)
        return False
    
    def sum_node(self):
        total = self.data
        if self.left:
            total += self.left.sum_node()
        if self.right:
            total += self.right.sum_node()
        return total

    def find_max(self):
        if self.right:
            return self.right.find_max()
        return self.data
    
    def find_min(self):
        if self.left:
            return self.left.find_min()
        return self.data
    
    def print_even(self):
        ds = []
        if self.data % 2 == 0:
            ds.append(self.data)
        if self.left:
            self.left.print_even()
        if self.right:
            self.right.print_even()
    
    def print_odd(self):
        ds = []
        if self.data % 2 != 0:
            ds.append(self.data)
        if self.left:
            self.left.print_odd()
        if self.right:
            self.right.print_odd()
    
    def KiemTraSNT(self):
        pass

    def print_SNT(self):
        pass

    def add_data(self, path):
        f = open("D:/Hoc python/DSA_24735131/test.txt", "r")
        s = f.read()
        arr = s.split()
        print(arr)
        bft = Node(int(arr[0]))
        for i in arr[1:]:
            bft.them_node(int(i))
```
