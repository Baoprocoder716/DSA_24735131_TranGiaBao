```python
def Tong(a,b):
    return a+b
def Hieu(a,b):
    return a-b
def Tich(a,b):
    return a*b
def Thuong(a,b):
    return a/b
```

```python
def menu():
    print("1.Tinh tong 2 so.")
    print("2.Tinh hieu 2 so.")
    print("3.Tinh tich 2 so.")
    print("4.Tinh thuong 2 so.")
    print("0.Thoat.")
```

```python
def main():
    menu()

    try:
        choose = input("Nhap menu:")
        if choose in ["1", "2", "3", "4"]:
            a = float(input("Nhap so thu nhat: "))
            b = float(input("Nhap so thu hai: "))
        choose = input("Nhap menu: ")
        if choose == "1":
            print("Tong la: ", Tong(a,b))
        elif choose == "2":
            print("Hieu la: ", Hieu(a,b))
        elif choose == "3":
            print("Tich la:", Tich(a,b))
        elif choose == "4":
            print("Thuong la:", Thuong(a,b))
        elif choose == "0":
            print("Exit")
        else:
            print("Vui long chon menu hop le.")
    except ValueError:
        print("Ban nhap so phu hop.")
    except Exception as e:
        print(e)
```

```python
main()
```
