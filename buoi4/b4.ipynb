```python
class SinhVien:
    def __init__(self, id, name, age=None, gpa=None):
        self.id = id
        self.name = name
        self.age = age
        self.gpa = gpa

    def InThongTin(self):
        return f"id: {self.id}, name:{self.name}, age:{self.age}, gpa:{self.gpa}"
```

```python
class SinhVienTaiChuc(SinhVien):
    def __init__(self, id, name, age=None, gpa=None, chuyen_nganh=None):
        super().__init__(id, name, age, gpa)
        self.chuyen_nganh = chuyen_nganh
    def InThongTin(self):
        return f"{super().InThongTin()} - Noi dao tao: {self.chuyen_nganh}"
```

```python
class SinhVienChinhQuy(SinhVien):
    def __init__(self, id, name, age=None, gpa=None, noi_dao_tao=None):
        super().__init__(id, name, age, gpa)
        self.noi_dao_tao = noi_dao_tao
    def InThongTin(self):
        return f"{super().InThongTin()} - Noi dao tao: {self.noi_dao_tao}"
```

```python
from abc import ABC, abstractmethod
class BaseSV(ABC):
    @abstractmethod
    def is_empty(self):
        pass
    @abstractmethod
    def push(self, sv):
        pass
    @abstractmethod
    def pop(self):
        pass
```

```python
class StackSV(BaseSV):
    def __init__(self):
        self.lst_stack = []
    def is_empty(self):
        return len(self.lst_stack) == 0
    def push(self, sv):
        self.lst_stack.append(sv)
    def pop(self):
        #if self.lst_stack.is_empty():
        #    return None
        #return self.lst_stack.pop()
        return None if self.lst_stack.is_empty() else self.lst_stack.pop()
```

```python
class QueueSV(BaseSV):
    def __init__(self):
        self.lst_queue = []
    def is_empty(self):
        return len(self.lst_queue) == 0
    def push(self, sv):
        self.lst_queue.append(sv)
    def pop(self):
        #if self.lst_queue.is_empty():
        #    return None
        #return self.lst_queue.pop(0)
        return None if self.lst_queue.is_empty() else self.lst_queue.pop(0)
```
